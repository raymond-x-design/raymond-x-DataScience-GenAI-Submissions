### Visualising the Data

**Task 8 — Bar chart.** Create a bar chart comparing the **average salary in USD** across experience levels (`EN`, `MI`, `SE`, `EX`). Add clear axis labels and a title. Which group has the highest average?

```python
# Calculate average salary in USD across experience levels
avg_salary = data.groupby('experience_level')['salary_in_usd'].mean().reindex(['EN', 'MI', 'SE', 'EX'])

# Create the bar chart
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_salary.index, y=avg_salary.values, palette='viridis')

# Add labels and title
plt.xlabel('Experience Level')
plt.ylabel('Average Salary in USD')
plt.title('Average Salary in USD by Experience Level')
plt.show()
```

**Task 9 — Scatter plot.** Make a scatter plot of `work_year` against `salary_in_usd`. Label the axes. What patterns do you notice? Are any points difficult to distinguish because they overlap?

```python
# Create a scatter plot of work_year against salary_in_usd
plt.figure(figsize=(8, 6))
sns.scatterplot(data=data, x='work_year', y='salary_in_usd', alpha=0.5)

# Add labels and title
plt.xlabel('Work Year')
plt.ylabel('Salary in USD')
plt.title('Scatter Plot of Work Year vs Salary in USD')
plt.xticks([2020, 2021, 2022, 2023, 2024])
plt.show()
```

There are a lot of overlapping points across all the years. There are also several outlier results that stray from the main overlapping ones.

**Task 10 — Distribution.** Plot the distribution of `salary_in_usd`. Does it look approximately normal, skewed, or something else? Are there extreme values?

```python
# Plot the distribution of salary_in_usd
plt.figure(figsize=(10, 6))
sns.displot(data['salary_in_usd'], kde=True, bins=40, color='skyblue')

# Add labels and title
plt.xlabel('Salary in USD')
plt.ylabel('Count')
plt.title('Distribution of Salary in USD')
plt.show()
```

The distribution seems to be slightly skewed towards the left. There doesn't seem to be many extreme factors either.

**Task 11 — Pairwise relationships.** Visualize pairwise relationships among `salary_in_usd`, `work_year`, and `remote_ratio`, including distributions of the individual variables.

Try to compare the variables in one combined visualization. What relationships or patterns do you notice?

```python
# Create a pairplot to visualize pairwise relationships and individual distributions
sns.pairplot(data[['salary_in_usd', 'work_year', 'remote_ratio']], diag_kind='kde', height=2.5)
plt.suptitle('Pairwise Relationships & Distributions', y=1.02)
plt.show()
```

It seems that they all have relationships with each other, except for work_year and remote_ratio.
Meanwhile, the other variables do share relationships with each other as shown by their distributions having overlapping variables, and outlier results
